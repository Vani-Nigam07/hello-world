```python
##  use dictionaries to count the number of items in a list

list = [1,1,2,2,3,4,5,5,5,6,7,7,8,8,9,10,10]
dict = {}

for x in list:
    if x in dict:
        dict[x] += 1
    else:
        dict[x] = 1

for key in dict:
    print(f"{key} occurs {dict[key]} times")
        
    
```

```python
import math
def calculate_lift_rounds(n, capacity):
    """
    Function to calculate the number of rounds the lift needs to cover.
    
    Parameters:
    n (int): Total number of people.
    capacity (int): Maximum number of people the lift can carry in one round.
    
    Returns:
    int: The number of rounds required to transport all people to the top floor.
    """
    # Your code here
    if n>capacity:
        if (n%capacity!=0):
            num_of_lifts = math.floor(n/capacity)+1
        else:
            num_of_lifts = n/capacity
    else:
        num_of_lifts = 1

    return num_of_lifts    
```

```python
calculate_lift_rounds(4, 3)  # Example usage
```

```python
def max_consecutive_difference(lst):
    max_diff = 0
    for i in range(len(lst)-1):
        diff = abs(lst[i] - lst[i+1])
        if diff > max_diff:
            max_diff = diff 
    return max_diff
```

```python
def merge_two_sorted_lists(list1, list2):
    for i in range(len(list2)):
        if list2[i] not in list1:
            list1.append(list2[i])
    list1.sort()
    return list1
```

```python
merge_two_sorted_lists(list1, [5,6,7,9,10])  # Example usage
```

```python
def rotate_list(lst, k):
    result = []
    n = len(lst)
    for i in range(len(lst)):
        if k!=0:
            result.append(lst[n-k])
            k-=1
        else:
            result.append(lst[n-i-len(result)])
    return result        

```

```python
rotate_list([1,2,3,4,5], 2)  
```

```python

def twoSum( nums, target):
    for i in range(len(nums)):
        if (target - nums[i]) in nums:
            return [i,nums.index(target- nums[i])]
```

```python
twoSum( [2,7,11,15], 9)  # Example usage
```
